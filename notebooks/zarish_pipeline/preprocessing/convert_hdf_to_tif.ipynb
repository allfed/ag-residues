```python
import os
from osgeo import gdal
```

```python
# Input
hdf_file = "../../../data/raw/MCD12C1.2010.hdf"

#Output
tif_file = "../../../data/raw/MCD12C1-confidence.2010.tif"  
```

```python
def hdf_to_tif(hdf_file, subdataset_index, tif_file):
    # Open the HDF file
    hdf_ds = gdal.Open(hdf_file, gdal.GA_ReadOnly)
    
    # List all subdatasets in the HDF file
    subdatasets = hdf_ds.GetSubDatasets()

    # Check if the subdataset index is valid
    if subdataset_index >= len(subdatasets):
        print("Invalid subdataset index.")
        return

    # Select the desired subdataset
    subdataset = subdatasets[subdataset_index][0]
    
    # Open the desired subdataset
    sds_ds = gdal.Open(subdataset, gdal.GA_ReadOnly)

    # Convert the subdataset to a TIFF file
    driver = gdal.GetDriverByName('GTiff')
    driver.CreateCopy(tif_file, sds_ds)

    # Close datasets
    sds_ds = None
    hdf_ds = None

    print(f"Converted {hdf_file} (subdataset {subdataset_index}) to {tif_file}")
```

```python
hdf_to_tif(hdf_file, 1, tif_file)
```
