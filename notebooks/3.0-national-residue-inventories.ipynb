```python
def get_dry_crop_residue_mass(crop_name, crop_production, lookup):
    if crop_name in lookup.index:
        return lookup.loc[crop_name].sg_ratio * crop_yield * lookup.loc[crop_name].res_dm_ratio
    return 0
```

```python
for crop_name, lookup_row in lookup.iterrows():
    national_crop_boundaries[f"{crop_name}_dry_residues"] = national_crop_boundaries[f"{crop_name}_sum"].apply(lambda s: get_dry_crop_residue_mass(crop_name, s, lookup))
```
